```python
import pandas as pd
import numpy as np
from matplotlib.pyplot import subplots
```

```python
Auto = pd.read_csv("Auto.csv", na_values=["?"])
Auto_new = Auto.dropna()
Auto_re = Auto_new.set_index("name")
```

```python
Auto_Range_Values = {}
for column in Auto_re:
    Auto_Range_Values[column] = Auto_re[column].max() - Auto_re[column].min()
print(Auto_Range_Values)
```

```python
Auto_STD_Values = {}
for column in Auto_re:
    Auto_STD_Values[column] = Auto_re[column].std().round()
print(Auto_STD_Values)
```

```python

for i in range(10, 85):
    Auto_re = Auto_re.drop(Auto_re.index[i])
```
