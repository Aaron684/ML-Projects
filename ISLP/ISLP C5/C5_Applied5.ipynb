```python
Boston = load_data('Boston')
mu_hat = Boston['medv'].mean()
mu_hat
```

```python
medv = Boston['medv']
se_mu_hat = np.std(medv, ddof=1) / np.sqrt(len(medv))
se_mu_hat
```

```python
medv = Boston['medv'].values

n_bootstrap = 1000
np.random.seed(42)

bootstrap_means = []
for _ in range(n_bootstrap):
    sample = np.random.choice(medv, size=len(medv), replace=True)
    bootstrap_means.append(np.mean(sample))

bootstrap_se = np.std(bootstrap_means)
print(f"Bootstrap Standard Error of the Mean: {bootstrap_se:.3f}")
```

```python
bootstrap_ci = np.percentile(bootstrap_means, [2.5, 97.5])

mean_medv = np.mean(medv)
std_error = np.std(medv, ddof=1) / np.sqrt(len(medv))
two_se_ci = [mean_medv - 2 * std_error, mean_medv + 2 * std_error]

print(f"Bootstrap 95% CI: {bootstrap_ci}")
print(f"Two-SE Rule 95% CI: {two_se_ci}")
```

```python
np.random.seed(43)

bootstrap_medians = []
for _ in range(n_bootstrap):
    sample = np.random.choice(medv, size=len(medv), replace=True)
    bootstrap_medians.append(np.median(sample))

bootstrap_med_se = np.std(bootstrap_medians)
print(f"Bootstrap Standard Error of the Median: {bootstrap_med_se:.3f}")
```

```python
np.random.seed(44)

bootstrap_tens = []
for _ in range(n_bootstrap):
    sample = np.random.choice(medv, size=len(medv), replace=True)
    bootstrap_tens.append(np.percentile(sample, 10))

bootstrap_tens_se = np.std(bootstrap_tens)
print(f"Bootstrap Standard Error of the 10th percentile: {bootstrap_tens_se:.3f}")
```
