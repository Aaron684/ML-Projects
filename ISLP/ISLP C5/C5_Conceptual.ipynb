```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_values = np.arange(1, 100001)
```

```python
probabilities = 1 - (1 - 1/n_values)**n_values
```

```python
plt.figure(figsize=(10, 6))
plt.plot(n_values, probabilities, color='blue')
plt.title("Probability of Inclusion in Bootstrap Sample vs Sample Size", fontsize=14)
plt.xlabel("Sample Size (n)", fontsize=12)
plt.ylabel("Inclusion Probability", fontsize=12)
plt.xscale('log')
plt.grid(True)
plt.tight_layout()
```

Probability of inclusion approaches 0.632, so one can likely assume the chance of an observation being in a bootstrap is this value

3. kfold cross validation is implemented by dividing the data randomly into equal k parts and using each part to estimate
the accuracy of the model. Each part is used once as a test set while the remaining parts are used for training, and the
mean is taken of the errors to calculate the final estimate. This is repeated k times

Kfold is usually more accurate than validation as it lowers bias and variance though it takes longer to compute. Compared to LOOCV, it is less accurate though much faster, especially on more complex models and larger datasets.

Most likely we would use bootstrap method to make an estimate on the standard deviation, though for models like linear regression there are equations to find it
