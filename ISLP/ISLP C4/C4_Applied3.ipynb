```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def Power3(x, a):
    return x**a
```

```python
Power3(3, 8)
```

```python
print(str(Power3(10, 3)) + " " + str(Power3(8, 17)) + " " + str(Power3(131, 3)))
```

```python
x = np.arange(1, 11)
y = Power3(x, 2)
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, marker='o', linestyle='-', color='teal')
ax.set_xlabel('x values')
ax.set_ylabel('f(x) = x^2')
ax.set_title('Plot of f(x) = x² Using Power3 Function');
```

```python
def PlotPower(x1, x2):
    c = np.arange(x1, x2 + 1)
    d = Power3(c, 2)
    fig, ax = plt.subplots()
    ax.plot(c, d, marker='o', linestyle='-', color='teal')
    ax.set_title('Plotted Power at Fixed Range');
```

```python
PlotPower(1, 10)
```
