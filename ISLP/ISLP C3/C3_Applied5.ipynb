```python
x = np.random.normal(loc=0, scale=1, size=100)
```

Using the normal() method, create a vector, eps, containing 100 observations drawn from a N(0,0.25) distribution—a normal
distribution with mean zero and variance 0.25.


```python
eps = np.random.normal(loc=0, scale=0.5, size=100)
```

```python
y = -1 + 0.5 * x + eps
```

```python
#B0 is 1, B1 is 0.5, and Y's length is 100
```

```python
plt.scatter(x, y);
```

```python
dfx = pd.DataFrame({'intercept': np.ones(len(x)), 'x': x})
```

```python
model = sm.OLS(y, dfx)
results = model.fit()
```
