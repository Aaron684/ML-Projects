```python
def getScores():
    scores = {}
    while True:
        studentID = input("Enter ID: ")
        studentScores = input("Enter list of scores: ").split(", ")
        total = 0
        if studentID == "":
            break
        else:
            for i in range(len(studentScores)):
                total += int(studentScores[i])
            scores[studentID] = total / len(studentScores)
    return scores
x = getScores()
print(x)
```
